## 1. Core Intuition: Why do we need Attention?

- **The Base Embedding ($E$):** Captures the general representation learned across training (e.g., `bank` is associated with money, loans, rivers, and shores). However, at lookup time, `emb('bank')` is identical regardless of the sentence.
- **The Problem:** In *"The river bank overflowed"*, `bank` refers to a water shore. In *"The money bank account"*, `bank` refers to a financial institution. The base embedding alone cannot distinguish between these contexts.
- **The Solution (Attention):** Attention allows `bank` to inspect its surrounding neighbors (`river` vs `money`) and blend their information into a new **Contextual Representation ($H$)**.

--- 
## 2. The Q, K, V Roles & Intuition

Each token embedding $X$ is projected into 3 separate vectors using shared weight matrices $W_Q, W_K, W_V$:

| Vector | Formula | Intuition / Question Asked |
    | :--- | :--- | :--- |
    | **Query ($Q$)** | $Q = X \cdot W_Q$ | **"What am I searching for?"** (The target token asks for context) |
    | **Key ($K$)** | $K = X \cdot W_K$ | **"What category/information do I contain?"** (Neighbor tokens advertise their content) |
    | **Value ($V$)** | $V = X \cdot W_V$ | **"What actual information can I pass along?"** (Content payload to be mixed) |

> **Crucial Rule:** The weight matrices $W_Q, W_K, W_V$ are **SHARED across all tokens** in the sequence. Each token gets its own $Q, K, V$ vectors because its input vector $X$ is different!

# PART 1: Forward Pass Mechanics in Pure NumPy

```python
import numpy as np

np.random.seed(42)
print(f"NumPy Version: {np.__version__}")
```

### Step 1: Input Sequence Matrix ($X$)

Sentence: **"The bank river overflowed"**
- Sequence length ($N$) = 4 tokens (`['The', 'bank', 'river', 'overflowed']`)
- Embedding dimension ($d_{\text{emb}}$) = 8

```python
tokens = ['The', 'bank', 'river', 'overflowed']
seq_len = len(tokens)
emb_dim = 8

# Sequence Matrix X (shape: 4 tokens x 8 dimensions)
X = np.random.randn(seq_len, emb_dim)
print(f"Sequence Matrix X shape: {X.shape} (4 tokens, 8 embedding dims)")
```

### Step 2: Linear Projections ($W_Q, W_K, W_V$) & Computing $Q, K, V$

- Shapes: $W_Q, W_K, W_V \in \mathbb{R}^{8 \times 4}$ (`emb_dim=8`, `head_dim=4`).

```python
head_dim = 4  # Dimension d_k

# Shared Projection Weight Matrices initialized randomly
W_Q = np.random.randn(emb_dim, head_dim) * 0.1
W_K = np.random.randn(emb_dim, head_dim) * 0.1
W_V = np.random.randn(emb_dim, head_dim) * 0.1

# Project sequence matrix X to get Q, K, V: (4, 8) @ (8, 4) -> (4, 4)
Q = X @ W_Q
K = X @ W_K
V = X @ W_V

print(f"Q shape: {Q.shape} | K shape: {K.shape} | V shape: {V.shape}")
```

### Step 3: Raw Attention Scores ($S = Q \cdot K^T$)

- Multiplying $Q$ `(4, 4)` by $K^T$ `(4, 4)` computes dot-product similarities between all token pairs.

```python
raw_scores = Q @ K.T
print("Raw Scores Matrix S (shape: 4, 4):")
print(raw_scores.round(3))
```

### Step 4: Why do we divide by $\sqrt{d_k}$?

Let's demonstrate **why scaling by $\sqrt{d_k}$ is necessary**! Without scaling, as dimension $d_k$ grows large, dot products become huge, causing `softmax` to push probabilities to 1.0 or 0.0 with near-zero gradients (vanishing gradient problem).

```python
def softmax(logits, axis=-1):
    e = np.exp(logits - np.max(logits, axis=axis, keepdims=True))
    return e / np.sum(e, axis=axis, keepdims=True)

d_k = head_dim
# 1. Unscaled Softmax
unscaled_attn = softmax(raw_scores * 10.0)  # Simulating unscaled large dot-products
# 2. Scaled Softmax
scaled_scores = raw_scores / np.sqrt(d_k)
attention_weights = softmax(scaled_scores)

print("=== Demonstration of Scaling ===")
print("Unscaled (Saturated) Softmax row 1 ('bank'):", unscaled_attn[1].round(4))
print("Scaled Softmax row 1 ('bank'):", attention_weights[1].round(4))
print("\n-> Notice how scaling prevents extreme 0 or 1 values, allowing healthy gradient flow!")
```

### Step 5: Weighted Sum of Values ($H = A \cdot V$)

Multiplying Attention Weights $A$ `(4, 4)` by Values $V$ `(4, 4)` produces the final **Contextual Representation Matrix $H$**.

```python
H = attention_weights @ V

print(f"Contextual Matrix H shape: {H.shape}")
print(f"Base vector for 'bank' (row 1 of X, first 4 dims): {X[1, :4].round(3)}")
print(f"New Contextual vector for 'bank' (row 1 of H):      {H[1].round(3)}")
```

# PART 2: Training the Network (Backprop & SGD in Pure NumPy)

Up to this point, $W_Q, W_K, W_V$ were random. Now we will train them on a classification task so the network **learns** to adjust its attention weights!

```python
# Vocabulary & Dataset
vocab = {"river": 0, "bank": 1, "overflowed": 2, "money": 3, "account": 4}
vocab_size = len(vocab)
emb_dim = 8
head_dim = 4

# Lookup Embedding Table E
E = np.random.randn(vocab_size, emb_dim) * 0.1

# Dataset: (Token IDs, Label)
# Label 0 = Nature/Water, Label 1 = Finance
dataset = [
    ([vocab["river"], vocab["bank"], vocab["overflowed"]], 0),  # Sentence 1
    ([vocab["money"], vocab["bank"], vocab["account"]], 1)      # Sentence 2
]

# Initialize Trainable Parameters
W_Q = np.random.randn(emb_dim, head_dim) * 0.1
W_K = np.random.randn(emb_dim, head_dim) * 0.1
W_V = np.random.randn(emb_dim, head_dim) * 0.1

# Classifier Layer (head_dim -> 2 classes)
W_c = np.random.randn(head_dim, 2) * 0.1
b_c = np.zeros((1, 2))

print("Network parameters initialized for training.")
```

### Step 7: Forward Pass & Backward Pass Functions

Matrix Gradient Equations derived from Backpropagation:
1. $\delta_{\text{logits}} = \text{probs} - y_{\text{onehot}}$
2. $\frac{\partial L}{\partial W_c} = h_{\text{bank}}^T \cdot \delta_{\text{logits}}$
3. $\frac{\partial L}{\partial H} = \delta_{\text{logits}} W_c^T$
4. $\frac{\partial L}{\partial V} = A^T \cdot \frac{\partial L}{\partial H}$
5. $\frac{\partial L}{\partial A} = \frac{\partial L}{\partial H} \cdot V^T$
6. $\delta_S = \frac{1}{\sqrt{d_k}} A \odot \left( \delta_A - \sum (\delta_A \odot A) \right)$
7. $\frac{\partial L}{\partial W_Q} = X^T \cdot (\delta_S K), \quad \frac{\partial L}{\partial W_K} = X^T \cdot (\delta_S^T Q), \quad \frac{\partial L}{\partial W_V} = X^T \cdot (A^T \frac{\partial L}{\partial H})$

```python
def forward(token_ids, W_Q, W_K, W_V, W_c, b_c):
    X = E[token_ids]
    Q = X @ W_Q
    K = X @ W_K
    V = X @ W_V
    
    d_k = head_dim
    S = (Q @ K.T) / np.sqrt(d_k)
    A = softmax(S, axis=-1)
    H = A @ V
    
    # Select 'bank' vector (index 1)
    h_bank = H[1:2, :]
    logits = h_bank @ W_c + b_c
    probs = softmax(logits, axis=-1)
    
    cache = (X, Q, K, V, S, A, H, h_bank, logits, probs)
    return probs, cache

def backward(probs, label, cache, W_Q, W_K, W_V, W_c):
    X, Q, K, V, S, A, H, h_bank, logits, probs = cache
    d_k = head_dim
    
    y_onehot = np.zeros((1, 2))
    y_onehot[0, label] = 1.0
    
    # Derivatives
    d_logits = probs - y_onehot
    dW_c = h_bank.T @ d_logits
    db_c = d_logits
    
    dH = np.zeros_like(H)
    dH[1:2, :] = d_logits @ W_c.T
    
    dV = A.T @ dH
    dA = dH @ V.T
    
    sum_dA_A = np.sum(dA * A, axis=-1, keepdims=True)
    dS = (A * (dA - sum_dA_A)) / np.sqrt(d_k)
    
    dQ = dS @ K
    dK = dS.T @ Q
    
    dW_Q = X.T @ dQ
    dW_K = X.T @ dK
    dW_V = X.T @ dV
    
    return dW_Q, dW_K, dW_V, dW_c, db_c
```

### Step 8: Training Loop with Pure NumPy SGD

```python
learning_rate = 0.1
epochs = 200

print("Starting SGD Training Loop in Pure NumPy...\n")

for epoch in range(1, epochs + 1):
    total_loss = 0.0
    for token_ids, label in dataset:
        # 1. Forward Pass
        probs, cache = forward(token_ids, W_Q, W_K, W_V, W_c, b_c)
        
        # 2. Cross-Entropy Loss
        loss = -np.log(probs[0, label] + 1e-12)
        total_loss += loss
        
        # 3. Backward Pass
        dW_Q, dW_K, dW_V, dW_c, db_c = backward(probs, label, cache, W_Q, W_K, W_V, W_c)
        
        # 4. SGD Parameter Update: W = W - lr * dW
        W_Q -= learning_rate * dW_Q
        W_K -= learning_rate * dW_K
        W_V -= learning_rate * dW_V
        W_c -= learning_rate * dW_c
        b_c -= learning_rate * db_c
        
    if epoch % 40 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d} | Total Loss: {total_loss:.4f}")
```

### Step 9: Inspecting the Trained Attention Weights!

Let's see how the trained $W_Q, W_K, W_V$ matrices cause `'bank'` to attend differently depending on context!

```python
print("\n================ FINAL TRAINED ATTENTION WEIGHTS ================\n")

# Test Sentence 1: "river bank overflowed"
probs1, cache1 = forward(dataset[0][0], W_Q, W_K, W_V, W_c, b_c)
attn1 = cache1[5]
pred1 = np.argmax(probs1)

print("Sentence 1: ['river', 'bank', 'overflowed']")
print(f"Predicted Class: {pred1} (0 = Nature/Water)")
print("Attention Weights for 'bank' (Row 1 -> ['river', 'bank', 'overflowed']):")
print(attn1[1].round(4))
print("-" * 60)

# Test Sentence 2: "money bank account"
probs2, cache2 = forward(dataset[1][0], W_Q, W_K, W_V, W_c, b_c)
attn2 = cache2[5]
pred2 = np.argmax(probs2)

print("Sentence 2: ['money', 'bank', 'account']")
print(f"Predicted Class: {pred2} (1 = Finance)")
print("Attention Weights for 'bank' (Row 1 -> ['money', 'bank', 'account']):")
print(attn2[1].round(4))
```
